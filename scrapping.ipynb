```python
import requests 
from bs4 import BeautifulSoup

url = "https://quotes.toscrape.com"
response = requests.get(url)#downloads website html
print(response.status_code)

#BeautifulSoup converts html in readable form
soup = BeautifulSoup(response.text,'html.parser')

quotes = soup.find_all("div",class_='quote')

for quote in quotes:
    text = quote.find('span',class_='text').text
    author = quote.find('small',class_='author').text
    print(text)
    print(author)
```

```python
import pandas as pd
url = "https://www.flipkart.com/search?q=mobile+phones&otracker=search&otracker1=search&marketplace=FLIPKART&as-show=off&as=off"

headers = {
    "User-Agent":(
        "Mozilla/5.0"
    )
}

response = requests.get(url,headers = headers)
print("Status code:",response.status_code)

if response.status_code==403:
    print("Website is blocking the request.Please try Again later or use a different User_Agent")
else:
    response.raise_for_status()
    soup = BeautifulSoup(response.text,"html.parser")
    print("Page loaded successfully")
    
    #parent product container
products = soup.find_all("div", clas_="lvjbLV")

print("Total Products Found:", len(products))

#final Data List
data = []

#loop through each product
for product in products:
      
      # Products Name
      name = product.find("div",class_="RG5Slk")
      
      if name:
          name = name.text.strip()
      else:
          name = "N/A"
      
      #Price
      price = product.find("div", class_="hZ3P6w DeU9vF")
      
      if price:
          price = price.text.strip()
      else:
          price = "N/A"
          
    #Rating
      rating = product.find("div",class_="MKif56")
    
      if rating:
        rating = rating.text.strip()
      else:
        rating = "N/A"
        
    #Product Link
      link = product.find('div',class_="k7wcnx")
         
      if link :
          link = "https://www.flipkart.com" + link.get("href")
      else:
          link = "N/A"
    #Store in list
      data.append([name,price,rating,link])
    
#DataFrame
df = pd.DataFrame(
    data,
    columns=["Product Name","Price","Rating","Product Link"]
)

#Display output
print(df)

#CSV Save
df.to_csv("flipkart_products.csv",index=False)
        
print("Data Saved Successfully")
            
```
