```python
import cv2
```

```python
cap = cv2.VideoCapture(0)#to setup the camera connection
while True:
    isPresent, frame = cap.read()#captures a single frame from the image
    cv2.imshow('my image',frame)
    
    if cv2.waitKey(1)& 0xFF==ord ('q'):
        break
cv2.destoryALLWindows
cap.release()
        
```
