```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('iris')
df
```

```python
#histogram - distribution of sepal length
plt.hist(df['sepal_length'],color = 'skyblue',edgecolor = 'black')
plt.title('distribution of sepal length')
plt.xlabel('sepal length')
plt.ylabel('frequency')
plt.show()
```

```python
plt.plot(df['petal_length'],color = 'orange')
plt.title('petal length of first 50 samples')
plt.xlabel('sample index')
plt.ylabel('petal length')
plt.show()
```

```python
#line plot - petal of first 50 samples
plt.plot(df['petal_length'][:50],color = 'orange')
plt.title('petal length of first 50 samples')
plt.xlabel('sample index')
plt.ylabel('petal length')
plt.show()
```

```python
#calculate the mean value of species
mean_value = df.groupby("species").mean()
mean_value
```

```python
mean_value.plot(kind='bar',figsize=(5,5))
plt.title("Average feature values per species")
plt.ylabel("average value")
plt.xticks(rotation=20)
plt.show()
```

```python
fig,ax = plt.subplots(figsize = (8,3))
#settings color figure and axes
fig.set_facecolor('skyblue')
ax.set_facecolor('lightyellow')
#setting the title and labels
ax.set_title('customized line plot ', color='red',fontsize=18,
             fontfamily='serif',fontweight='bold',loc='right')
ax.set_xlabel('X-axis',color='green',fontsize=15,fontfamily='serif')
ax.set_ylabel('Y-axis',color='green',fontsize=15,fontfamily='serif')
#data plotting
x=[1,2,3,4,5,6,7,8,9]
y=[3,6,2,5,7,8,9,4,1]
ax.plot(x,y,
        color='red',
        linewidth='1',
        linestyle='--',
        marker='o',
        markersize=6,
        markerfacecolor='yellow',
        markeredgecolor='black',
        markeredgewidth=0.5
        )
#adding text
ax.text(3,7,'sample text',fontsize=12,fontfamily='serif',color='blue',
        fontweight='bold')
        #setting partition
ax.axhline(y=5,color='black',alpha=0.4)
ax.axvline(x=5,color='black')
        
```
