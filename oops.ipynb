```python
#constructor with parameters
class Student:
    def __init__(self, name, age):
      self.name = name
      self.age = age
      
s1 = Student("john",14)
print(s1.name,s1.age)     
        
```

```python
#object and class with method
class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age
        
    def display(self):
        print(f"hello my name is {self.name} and i am {self.age}years old")
        print("Nice to meet you!")
            
s1 = Student("john",14)
s1.display()          
```
